```python
import pandas as pd
import numpy as np

df = pd.read_csv('railway.csv')
df.columns = [col.replace(' ', '_') for col in df.columns]

df['Railcard'] = df['Railcard'].fillna('None')
df['Reason_for_Delay'] = df['Reason_for_Delay'].fillna('No Delay')

delay_map = {
    'Signal failure':     'Signal Failure',
    'Weather Conditions': 'Weather',
    'Staff Shortage':     'Staffing',
}
df['Reason_for_Delay'] = df['Reason_for_Delay'].replace(delay_map)


same_time_mask = (
    (df['Arrival_Time'] == df['Actual_Arrival_Time']) &
    (df['Journey_Status'] == 'Delayed')
)
df.loc[same_time_mask, 'Journey_Status']   = 'On Time'
df.loc[same_time_mask, 'Reason_for_Delay'] = 'No Delay'
df.loc[same_time_mask, 'Refund_Request']   = 'No'


DateConvP = pd.to_datetime(df['Date_of_Purchase'], format='%m/%d/%Y')
DateConvJ = pd.to_datetime(df['Date_of_Journey'],  format='%m/%d/%Y')

date_str  = DateConvJ.dt.strftime('%Y-%m-%d')
dept      = pd.to_datetime(date_str + ' ' + df['Departure_Time'])
sched_arr = pd.to_datetime(date_str + ' ' + df['Arrival_Time'])
act_arr   = pd.to_datetime(date_str + ' ' + df['Actual_Arrival_Time'], errors='coerce')

sched_arr[sched_arr < dept] += pd.Timedelta(days=1)
act_arr[(act_arr.notnull()) & (act_arr < dept)] += pd.Timedelta(days=1)

df['Delay_Minutes'] = (act_arr - sched_arr).dt.total_seconds() / 60
df.loc[df['Journey_Status'] == 'On Time',   'Delay_Minutes'] = 0
df.loc[df['Journey_Status'] == 'Cancelled', 'Delay_Minutes'] = np.nan
df['Journey_Duration_Minutes'] = (sched_arr - dept).dt.total_seconds() / 60

def calc_refund(row):
    if row['Journey_Status'] == 'Cancelled' and row['Refund_Request'] == 'Yes':
        return row['Price']
    if row['Journey_Status'] == 'Delayed' and row['Refund_Request'] == 'Yes':
        mins = row['Delay_Minutes']
        if mins >= 120:   return row['Price']
        elif mins >= 60: return row['Price'] * 0.50
        elif mins >= 30: return row['Price'] * 0.25
        else:            return 0
    return 0

df['Refunded_Amount'] = df.apply(calc_refund, axis=1)
df['Profit'] = df['Price'] - df['Refunded_Amount']
df['Transaction_ID'] = range(len(df))
df['Journey_Date'] = DateConvJ.dt.date
df['Purchase_Date'] = DateConvP.dt.date
df['Time_ID'] = pd.to_datetime(df['Time_of_Purchase'], format='%H:%M:%S').dt.hour


all_dates = pd.to_datetime(pd.concat([DateConvP, DateConvJ]).unique())
dates = pd.DataFrame({'Full_Date': all_dates.date})
dates['Year'] = pd.to_datetime(dates['Full_Date']).dt.year
dates['Month_Num'] = pd.to_datetime(dates['Full_Date']).dt.month
dates['Month_Name'] = pd.to_datetime(dates['Full_Date']).dt.month_name()
dates['Day_Name'] = pd.to_datetime(dates['Full_Date']).dt.day_name()

status = pd.DataFrame({'Journey_Status': df['Journey_Status'].unique()}).assign(Status_ID=lambda x: range(len(x)))
routes = df[['Departure_Station', 'Arrival_Destination']].drop_duplicates()
routes['Route_Name'] = routes['Departure_Station'] + " to " + routes['Arrival_Destination']
routes['Route_ID'] = range(len(routes))

ticket_type = pd.DataFrame({'Ticket_Type': df['Ticket_Type'].unique()}).assign(Ticket_Type_ID=lambda x: range(len(x)))
ticket_class = pd.DataFrame({'Ticket_Class': df['Ticket_Class'].unique()}).assign(Ticket_Class_ID=lambda x: range(len(x)))
railcards = pd.DataFrame({'Railcard_Type': df['Railcard'].unique()}).assign(Railcard_ID=lambda x: range(len(x)))
payments = pd.DataFrame({'Payment_Method': df['Payment_Method'].unique()}).assign(Payment_ID=lambda x: range(len(x)))
purchases = pd.DataFrame({'Purchase_Type': df['Purchase_Type'].unique()}).assign(Purchase_ID=lambda x: range(len(x)))
refunds = pd.DataFrame({'Refund_Request': df['Refund_Request'].unique()}).assign(Refund_ID=lambda x: range(len(x)))
delay_reasons = pd.DataFrame({'Reason_for_Delay': df['Reason_for_Delay'].unique()}).assign(Delay_ID=lambda x: range(len(x)))

times = pd.DataFrame({'Hour_24': range(24)})
times['Time_Bucket'] = pd.cut(times['Hour_24'], bins=[-1, 6, 12, 18, 24], labels=['Night', 'Morning', 'Afternoon', 'Evening'])
times['Is_OffPeak'] = np.where(times['Hour_24'].isin([7,8,9,16,17,18]), 0, 1)

fact = df.copy()
fact = fact.merge(status, on='Journey_Status').merge(routes, on=['Departure_Station', 'Arrival_Destination'])
fact = fact.merge(ticket_type, on='Ticket_Type').merge(ticket_class, on='Ticket_Class')
fact = fact.rename(columns={'Railcard': 'Railcard_Type'}).merge(railcards, on='Railcard_Type')
fact = fact.merge(payments, on='Payment_Method').merge(purchases, on='Purchase_Type')
fact = fact.merge(refunds, on='Refund_Request').merge(delay_reasons, on='Reason_for_Delay')

fact = fact[['Transaction_ID', 'Journey_Date', 'Purchase_Date', 'Time_ID', 'Status_ID', 'Route_ID',
             'Ticket_Type_ID', 'Ticket_Class_ID', 'Railcard_ID', 'Payment_ID', 'Purchase_ID',
             'Refund_ID', 'Delay_ID', 'Price', 'Refunded_Amount', 'Profit',
             'Delay_Minutes', 'Journey_Duration_Minutes']]
```

```python
# Q1. Total number of journeys recorded
print("Total Trips:", len(fact))
```

```python
# Q2. How many journeys occur yearly, monthly, daily?
q2_data = fact.merge(dates, left_on="Journey_Date", right_on="Full_Date", how="left")

print("\n Yearly Journeys:")
yearly_journeys = q2_data.groupby("Year")["Transaction_ID"].count().reset_index().rename(columns={"Transaction_ID": "Total_Journeys"})
print(yearly_journeys.to_string(index=False))

print("\n Monthly Journeys:")
monthly_journeys = (q2_data.groupby(["Year", "Month_Num", "Month_Name"])["Transaction_ID"]
                    .count().reset_index().rename(columns={"Transaction_ID": "Total_Journeys"})
                    .sort_values(["Year", "Month_Num"]))
print(monthly_journeys.to_string(index=False))

print("\n Daily Journeys :")
daily_journeys = q2_data.groupby("Full_Date")["Transaction_ID"].count().reset_index().rename(columns={"Transaction_ID": "Total_Journeys"})
print(daily_journeys.head(10).to_string(index=False))
```

```python
# Q3. Distribution of journey statuses
trips_status_merged = fact.merge(status, on="Status_ID")
status_distribution = trips_status_merged.groupby("Journey_Status")["Transaction_ID"].count().reset_index().rename(columns={"Transaction_ID": "Trips"})
status_distribution["Percentage"] = (status_distribution["Trips"] / status_distribution["Trips"].sum() * 100).round(2)
print("\n Journey Status Distribution:")
print(status_distribution.sort_values("Trips", ascending=False).to_string(index=False))
```

```python
# Q4 & Q5. Averages
print(f"\n Avg Journey Duration: {fact['Journey_Duration_Minutes'].mean():.2f} minutes")
print(f"\n Avg Delay: {fact['Delay_Minutes'].mean():.2f} minutes")
```

```python
# Q6, Q7, Q8. Revenue
print(f"\n Total Revenue: £{fact['Price'].sum():,.2f}")
total_refunds = fact["Refunded_Amount"].sum()
print(f" Total Refunds: £{total_refunds:,.2f}")
print(f" Net Profit: £{fact['Profit'].sum():,.2f}")
```

```python
# Q9. Revenue by ticket type
revenue_by_ticket = fact.merge(ticket_type, on="Ticket_Type_ID").groupby("Ticket_Type").agg(
    Gross_Revenue=("Price", "sum"), Net_Revenue=("Profit", "sum"), Refunds=("Refunded_Amount", "sum")
).sort_values("Gross_Revenue", ascending=False).reset_index()
print("\n Revenue by Ticket Type:")
print(revenue_by_ticket.to_string(index=False))
```

```python
# Q10. Revenue by ticket class
revenue_by_class = fact.merge(ticket_class, on="Ticket_Class_ID").groupby("Ticket_Class").agg(
    Gross_Revenue=("Price", "sum"), Net_Revenue=("Profit", "sum"), Refunds=("Refunded_Amount", "sum")
).sort_values("Gross_Revenue", ascending=False).reset_index()
print("\n Revenue by Ticket Class:")
print(revenue_by_class.to_string(index=False))
```

```python
# Q11. How does revenue vary by month?
revenue_by_month = (fact.merge(dates, left_on="Purchase_Date", right_on="Full_Date", how="left")
                    .groupby(["Year", "Month_Num", "Month_Name"])
                    .agg(
                        Gross_Revenue=("Price", "sum"),
                        Net_Revenue=("Profit", "sum"),
                        Refunds=("Refunded_Amount", "sum")
                    )
                    .reset_index()
                    .sort_values(["Year", "Month_Num"]))

print("\n Revenue Variation by Month:")
print(revenue_by_month.to_string(index=False))
```

```python
# Q12. Revenue by route
revenue_by_route = fact.merge(routes, on="Route_ID").groupby("Route_Name").agg(
    Gross_Revenue=("Price","sum"), Net_Revenue=("Profit","sum"), Refunds=("Refunded_Amount","sum")
).sort_values("Gross_Revenue", ascending=False).reset_index()
print("\n. Revenue by Route:")
print(revenue_by_route.to_string(index=False))
```

```python
# Q13. Average ticket price per route
route_avg_price = fact.merge(routes, on="Route_ID").groupby("Route_Name")["Price"].mean().round(2).reset_index().rename(columns={"Price":"Avg_Price"})
print("\n Avg Ticket Price per Route:")
print(route_avg_price.sort_values("Avg_Price", ascending=False).to_string(index=False))
```

```python
# Q14. Railcard usage
railcard_data = fact.merge(railcards, on="Railcard_ID")
railcard_data["Railcard_Group"] = np.where(railcard_data["Railcard_Type"] == "None", "No Railcard", "Has Railcard")
railcard_rev = railcard_data.groupby("Railcard_Group").agg(
    Tickets=("Transaction_ID","count"), Avg_Price=("Price","mean"), Gross_Revenue=("Price","sum")
).reset_index()
print("\n Railcard Usage vs Revenue:")
print(railcard_rev.to_string(index=False))
```

```python
# Q15. Payment method frequency
payment_freq = fact.merge(payments, on="Payment_ID").groupby("Payment_Method")["Transaction_ID"].count().reset_index().rename(columns={"Transaction_ID":"Tickets"})
print("\n Payment Method Frequency:")
print(payment_freq.sort_values("Tickets", ascending=False).to_string(index=False))
```

```python
# Q16. Which payment method generates the highest revenue?
revenue_by_payment = (fact.merge(payments, on="Payment_ID")
                      .groupby("Payment_Method")
                      .agg(
                          Tickets=("Transaction_ID", "count"),
                          Avg_Price=("Price", "mean"),
                          Gross_Revenue=("Price", "sum"),
                          Net_Revenue=("Profit", "sum"),
                          Refunds=("Refunded_Amount", "sum")
                      )
                      .reset_index()
                      .sort_values("Gross_Revenue", ascending=False))

revenue_by_payment["Avg_Price"] = revenue_by_payment["Avg_Price"].round(2)

print("\n Revenue by Payment Method:")
print(revenue_by_payment.to_string(index=False))
```
